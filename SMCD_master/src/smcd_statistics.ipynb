```python
import numpy as np
import scipy.sparse as sp
import os

def sparse_to_edges(sparse_mx):
    """Return number of edges in a sparse adjacency matrix."""
    return sparse_mx.nnz  # non-zero entries = edges

def load_smcd_dataset(dataset):
    path = f"data/{dataset}/"
    
    # Load labels
    labels = np.load(os.path.join(path, "labels.npy")).astype(int)
    num_classes = len(np.unique(labels))
    
    # Load feature files that exist
    possible_feat_files = ["p_feat.npz", "a_feat.npz", "s_feat.npz", 
                           "c_feat.npz", "m_feat.npz", "d_feat.npz"]
    num_nodes = 0
    for f in possible_feat_files:
        f_path = os.path.join(path, f)
        if os.path.exists(f_path):
            feat = sp.load_npz(f_path)
            num_nodes += feat.shape[0]
    
    # Load metapaths (edges)
    mp_folder = os.path.join(path, "mp")
    edge_count = 0
    if os.path.exists(mp_folder):
        for mp_file in os.listdir(mp_folder):
            if mp_file.endswith(".npz"):
                adj = sp.load_npz(os.path.join(mp_folder, mp_file))
                edge_count += sparse_to_edges(adj)
    
    return num_nodes, edge_count, num_classes

# Example usage
for dataset in ["acm", "dblp", "imdb"]:
    n_nodes, n_edges, n_classes = load_smcd_dataset(dataset)
    print(f"{dataset.upper()}: Nodes={n_nodes}, Edges={n_edges}, Classes={n_classes}")
```
