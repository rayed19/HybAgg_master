```python
# R : generate the average results of EGAI F1 scores results 

import numpy as np

# Load the saved result array
result_array = np.load("result_array.npy")  # shape: (num_percents, num_ratios, num_seeds, 3)

percent_list = [0.8, 0.5, 0.2]

print("=== Mean and Std for Each Percent ===")
for per_index, percent in enumerate(percent_list):
    # Average over ratios and seeds for this percent
    mean_results = result_array[per_index, :, :, :].mean(axis=(0, 1))
    std_results  = result_array[per_index, :, :, :].std(axis=(0, 1))
    
    accuracy_mean, micro_f1_mean, macro_f1_mean = mean_results
    accuracy_std,  micro_f1_std,  macro_f1_std  = std_results
    
    print(f"Percent {percent} :")
  #  print("  Accuracy : {:.4f} ± {:.4f}".format(accuracy_mean, accuracy_std))
    print("  Micro-F1 : {:.4f} ± {:.4f}".format(micro_f1_mean, micro_f1_std))
    print("  Macro-F1 : {:.4f} ± {:.4f}".format(macro_f1_mean, macro_f1_std))
```

```python
# R : generate average results per percent AND per ratio

import numpy as np

# Load saved results
# Shape:
# (num_percents, num_ratios, num_seeds, 3)
result_array = np.load("result_array.npy")

percent_list = [0.8, 0.5, 0.2]
ratio_list   = [0.1, 0.5, 0.8]

print("=== Mean and Std for Each Percent and Ratio ===")

for per_index, percent in enumerate(percent_list):

    print("\n==============================")
    print(f"Percent {percent}")
    print("==============================")

    for ratio_index, ratio in enumerate(ratio_list):

        # Select all seeds for this percent + ratio
        # Shape becomes:
        # (num_seeds, 3)
        results = result_array[per_index, ratio_index]

        # Mean/std over seeds only
        mean_results = results.mean(axis=0)
        std_results  = results.std(axis=0)

        accuracy_mean, micro_f1_mean, macro_f1_mean = mean_results
        accuracy_std,  micro_f1_std,  macro_f1_std  = std_results

        print(f"\nRatio {ratio} :")

        print("  Accuracy  : {:.4f} ± {:.4f}".format(
            accuracy_mean, accuracy_std))

        print("  Micro-F1  : {:.4f} ± {:.4f}".format(
            micro_f1_mean, micro_f1_std))

        print("  Macro-F1  : {:.4f} ± {:.4f}".format(
            macro_f1_mean, macro_f1_std))
```
